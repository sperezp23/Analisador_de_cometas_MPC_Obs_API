```python
import requests

nombre_cometa = r'C/2023 A3'

response = requests.get("https://data.minorplanetcenter.net/api/get-orb", json={"desig": nombre_cometa})

if response.ok:
    mpc_orb = response.json()[0]['mpc_orb']
else:
    print("Error: ", response.status_code, response.content)


tp_jd_mpc = mpc_orb[0]['COM']['coefficient_values'][-1]
```

```python
import pandas as pd

mjd = 60580.7538482761

fecha = pd.Timestamp('1970-01-01') + pd.to_timedelta(mjd - 40587, unit='D')

print(fecha)
```

```python
import pandas as pd

def mjd_to_date(mjd):
    return (
        pd.Timestamp('1970-01-01')
        + pd.to_timedelta(mjd - 40587, unit='D')
    ).strftime('%Y-%m-%d')

mjd_to_date(tp_jd_mpc)
```

```python
import pandas as pd

def mjd_to_timestamp(mjd):
    return (pd.Timestamp('1970-01-01') + pd.to_timedelta(mjd - 40587, unit='D')).normalize()

fecha_perihelio = mjd_to_timestamp(60580.7538482761)

print(fecha_perihelio)
print(type(fecha_perihelio))
```
